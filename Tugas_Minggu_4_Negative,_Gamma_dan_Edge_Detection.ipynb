```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

# Load the image in color (RGB)
img_rgb = cv2.imread('/content/tomat.jpg')

# Convert the RGB image to grayscale
img_gray = cv2.cvtColor(img_rgb, cv2.COLOR_BGR2GRAY)

# Apply logarithmic transformation to the grayscale image
c = 255 / (np.log(1 + np.max(img_gray)))
log_image_gray = c * (np.log(img_gray + 1))
log_image_gray = np.array(log_image_gray, dtype=np.uint8)

plt.imshow(log_image_gray, cmap='gray')
plt.title('Citra Logaritmik (dari Grayscale)')
plt.show()
```

```python
gamma = 0.5
gamma_corrected = np.array(255 * (img / 255) ** gamma, dtype='uint8')
plt.imshow(gamma_corrected, cmap='gray')
plt.title('Citra Gamma')
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

# Load the image in grayscale
img = cv2.imread('/content/traffic sign.jpg', cv2.IMREAD_GRAYSCALE)

# Apply Sobel operator in x direction
sobelx = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=5)

# Apply Sobel operator in y direction
sobely = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=5)

# Compute the magnitude of the gradient
sobel = np.sqrt(sobelx**2 + sobely**2)

# Display the result
plt.imshow(sobel, cmap='gray')
plt.title('Edge Detection (Sobel)')
plt.show()
```
